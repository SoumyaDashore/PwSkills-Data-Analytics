### Q1. Fill missing values using the following:
- Age ---> Median
- Salary ---> Mean
- Hours_Worked_Per_Week ---> Median
- Performance_Score ---> Mean

Display the dataset after handling missing values.

````python
# Solution

import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()

df = pd.read_csv('employee_productivity_dataset.csv')

print("Original Dataset Shape:", df.shape)

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# Fill missing values
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

df['Hours_Worked_Per_Week'] = df['Hours_Worked_Per_Week'].fillna(
    df['Hours_Worked_Per_Week'].median()
)

df['Performance_Score'] = df['Performance_Score'].fillna(
    df['Performance_Score'].mean()
)

print("\nValues Used for Imputation:")
print("Age Median:", df['Age'].median())
print("Salary Mean:", df['Salary'].mean())
print("Hours Worked Median:", df['Hours_Worked_Per_Week'].median())
print("Performance Score Mean:", df['Performance_Score'].mean())

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDataset After Handling Missing Values:")
display(df)
````

### Q2. Convert these columns into numeric values using label Encoding:
- Gender
- Department

Show the updated columns.

````python
# Solution

from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder
gender_encoder = LabelEncoder()
department_encoder = LabelEncoder()

# Encode Gender
df['Gender'] = gender_encoder.fit_transform(df['Gender'])

# Encode Department
df['Department'] = department_encoder.fit_transform(df['Department'])

print("Gender Encoding:")
for label, value in zip(
    gender_encoder.classes_,
    gender_encoder.transform(gender_encoder.classes_)
):
    print(label, "->", value)

print("\nDepartment Encoding:")
for label, value in zip(
    department_encoder.classes_,
    department_encoder.transform(department_encoder.classes_)
):
    print(label, "->", value)

print("\nUpdated Gender and Department:")
display(df[['Gender', 'Department']].head(10))
````

### Q3. Apply One-Hot Encoding on:

````python
# Solution

# One-Hot Encoding
df_onehot = pd.get_dummies(
    df,
    columns=['Work_Mode', 'Location'],
    dtype=int
)

print("Dataset after One-Hot Encoding:")
display(df_onehot)

print("\nNew One-Hot Encoded Columns:")

new_columns = [
    col for col in df_onehot.columns
    if col.startswith('Work_Mode_') or col.startswith('Location_')
]

print(new_columns)

print("\nNumber of New Columns Created:", len(new_columns))
````

### Q7. Create a pipeline that:
- Applies encoding to categorical columns
- Applies scaling to numerical columns

Use:
- ColumnTransformer
- Pipeline

````python
# Solution

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical columns
numeric_columns = [
    'Age',
    'Hours_Worked_Per_Week',
    'Projects_Completed',
    'Performance_Score',
    'Salary'
]

# Categorical columns
categorical_columns = [
    'Gender',
    'Department',
    'Work_Mode',
    'Location'
]

# Numerical transformation
numeric_transformer = Pipeline(
    steps=[
        ('scaler', StandardScaler())
    ]
)

# Categorical transformation
categorical_transformer = Pipeline(
    steps=[
        (
            'encoder',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            )
        )
    ]
)

# ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_columns),
        ('cat', categorical_transformer, categorical_columns)
    ]
)

# Complete Pipeline
pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor)
    ]
)

print("Preprocessing Pipeline Created Successfully.")
````

### Q8. Apply the pipeline on the dataset.

Display:
- Transformed dataset
- Shape of final dataset

````python
# Solution

# Apply pipeline
X_transformed = pipeline.fit_transform(df)

# Get feature names
feature_names = pipeline.named_steps[
    'preprocessor'
].get_feature_names_out()

# Convert transformed array into DataFrame
transformed_df = pd.DataFrame(
    X_transformed,
    columns=feature_names
)

print("Transformed Dataset:")
display(transformed_df.head(10))

print("\nFinal Dataset Shape:")
print(transformed_df.shape)

print("\nNumber of Features:")
print(len(feature_names))
````

### Q9. Why is Scaling important in Python?

### Solution
Scaling is important in machine learning because different features can have very different numerical ranges.

For example, in this dataset:
```
Age                     → approximately 20–40
Salary                  → approximately 42,000–65,000
Hours Worked            → approximately 35–50
Projects Completed      → small integer values
```
If these values are used without scaling, the large numerical values such as `Salary` can have a disproportionately large effect on algorithms that depend on distances or feature magnitudes.

Scaling provides features with comparable numerical ranges.

**Main benefits**

1. Prevents large-scale features from dominating smaller-scale features.
2. Improves numerical stability.
3. Helps distance-based algorithms.
4. Helps optimization-based algorithms converge more efficiently.
5. Is particularly useful for algorithms such as:
    - K-Nearest Neighbors
    - K-Means
    - Support Vector Machines
    - Logistic Regression
    - Neural Networks

### Q10. Why do we convert categorical data into numerical form?

### Solution
Machine-learning algorithms generally operate on numerical data. Categorical values such as:
```
Gender = Male
Department = IT
Work_Mode = Remote
Location = Pune
```
cannot normally be directly processed as mathematical feature values.

Therefore, categorical variables are transformed into numerical representations.

For example, Label Encoding:
```
Female → 0
Male   → 1
```
One-Hot Encoding:
```
Work_Mode_Hybrid
Work_Mode_Onsite
Work_Mode_Remote
```
with values such as:
```
0  0  1
```
for a Remote employee.
