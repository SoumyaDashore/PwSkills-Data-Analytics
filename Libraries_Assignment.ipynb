### Q1. Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows along with the dataset information to understand its structure.

```python
# Solution
import pandas as pd
import numpy as np

from google.colab import files
uploaded = files.upload()

# Load the dataset
df = pd.read_csv('retail_dataset.csv')

# Display first 5 rows
print("First 5 rows of the dataset:")
print(df.head())

# Display dataset information
print("\nDataset information:")
print(df.info())
```

### Q2. Inspect the dataset and identify missing values in each column. Display the total number of missing values present in the dataset.

```python
# Solution

# Missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Total missing values in the dataset
print("\nTotal missing values in the dataset:", df.isnull().sum().sum())
```

### Q3. Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.

```python
# Solution

# Calculate mean value
quantity_mean = df['Quantity'].mean()
price_mean = df['Price'].mean()

print("Mean values:")
print("Quantity Mean:", quantity_mean)
print("Price Mean:", price_mean)

# Replace missing values with mean
df['Quantity'] = df['Quantity'].fillna(quantity_mean)
df['Price'] = df['Price'].fillna(price_mean)

# Check if missing values are replaced
print("\nMissing values after replacement:")
print(df[['Quantity', 'Price']].isnull().sum())
```

### Q4. Remove all rows where the Product Category or Region column contains missing values.

```python
# Solution

# Remove rows with missing Product Category or Region
df = df.dropna(subset=['Product Category', 'Region'])

# Reset index
df = df.reset_index(drop=True)

print("Dataset after removing missing values of Product Category or Region:")
print(df)

# Check if missing values are replaced
print("\nMissing values after replacement:")
print(df[['Product Category', 'Region']].isnull().sum())
```

### Q5. Create a new column called Revenue using NumPy, where:
Revenue = Quantity * Price
Display the updated dataset with the new column.

```python
# Solution

# Calculate Revenue using NumPy
df['Revenue'] = np.multiply(df['Quantity'], df['Price'])

print("Updated dataset with Revenue column:")
print(df)
```

### Q6. Using NumPy, calculate the total revenue generated from all transactions in the dataset.

```python
# Solution

# Calculate total revenue using NumPy
total_revenue = np.sum(df['Revenue'])

print("Total revenue generated from all transactions:", round(total_revenue, 3))
```

### Q7. Group the dataset by Product Category and calculate the total revenue generated by each category.

```python
# Solution

# Group by Product Category and calculate total revenue
revenue_by_category = df.groupby('Product Category')['Revenue'].sum()

print("Total revenue generated by each category:")
print(revenue_by_category)
```

### Q8. Based on the revenue calculated for each product category, identify the top 3 categories with the highest revenue and the bottom 3 categories with the lowest revenue.

```python
# Solution

# Sorting revenue_by_category in descending order
sorted_revenue_by_category = revenue_by_category.sort_values(ascending=False)

# Top 3 categories with highest revenue
top_3_categories = sorted_revenue_by_category.head(3)

# Bottom 3 categories with lowest revenue
bottom_3_categories = sorted_revenue_by_category.tail(3)

print("Top 3 categories with highest revenue:")
print(top_3_categories)

print("\nBottom 3 categories with lowest revenue:")
print(bottom_3_categories)
```

### Q9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the region with the highest revenue and the region with the lowest revenue.

```python
# Solution

# Group by Region and calculate total revenue
revenue_by_region = df.groupby('Region')['Revenue'].sum().sort_values(ascending=False)

# Region with highest revenue
highest_revenue_region = revenue_by_region.index[0]

# Region with lowest revenue
lowest_revenue_region = revenue_by_region.index[-1]

print(f"Total revenue generated in each region: {revenue_by_region}")
print(f"Region with highest revenue: {highest_revenue_region}")
print(f"Region with lowest revenue: {lowest_revenue_region}")
```

### Q10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column using NumPy.

```python
# Solution

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Extract month from Date column
df['Month'] = df['Date'].dt.month

# Group by month and calculate total monthly revenue
monthly_revenue = df.groupby('Month')['Revenue'].sum()

print("Total monthly revenue:")
print(monthly_revenue)

# Calculate statistics using NumPy
mean_revenue = np.mean(df['Revenue'])
median_revenue = np.median(df['Revenue'])
std_revenue = np.std(df['Revenue'])

print("\nStatistics of Revenue column:")
print("Mean:", mean_revenue, "\nMedian:", median_revenue, "\nStandard Deviation:", std_revenue)
```
