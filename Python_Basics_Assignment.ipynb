### Q2. Explain the difference between List and Tuple in Python.

#### Solution:
| List | Tuple |
| --- | --- |
| Lists are mutable | Tuples are immutable |
| created using `[]` | created using `()` |
| Elements can be added, removed, or changed | Elements cannot be changed after creation |
|Generally requires more memory | Generally requires less memory |
| Example `[1,2,3]` | Example `(1,2,3)` |

### Q5. Write a Python program to check whether a number is even or odd using conditional statements.

```python
# Solution
num = int(input("Enter a number: "))
if num % 2 == 0:
  print(num, "is even")
else:
  print(num, "is odd")
```

### Q7. Create a list of five numbers and print the maximum number from the list.

```python
# Solution
numbers = [10, 25, 7, 45, 18]
max_num = max(numbers)
print("The maximum number is", max_num)
```

### Q8. Write a program to remove duplicate values from a list using a set.

```python
# Solution
my_list = [1, 2, 2, 3, 4, 4, 5]
unique_list = list(set(my_list))
print(unique_list)
```

### Q9. Write a function that returns the square of a number.

```python
# Solution
def square(num):
  return num ** 2

result = square(141)
print("Square is: ", result)
```

### Q10. Write a program to count how many times a number appears in a list.
Example List

`[2,3,4,2,5,2]`

```python
# Solution
my_list = [2,3,4,2,5,2]
print(my_list)
num = int(input("Enter a number to count from the List:"))
count = my_list.count(num)
print(f"The number {num} appears {count} times.")
```
