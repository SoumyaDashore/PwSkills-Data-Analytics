### Q9. Generate summary statistics using describe() from the result identify:
- Mean sales value
- Maximum quantity sold
- Minimum sales value

```python
# Solution

print("Summary Statistics:")
display(df.describe())

# Mean sales value
print("\nMean Sales Value:", df["Sales"].mean())

# Maximum quantity sold
print("Maximum Quantity Sold:", df["Quantity"].max())

# Minimum sales value
print("Minimum Sales Value:", df["Sales"].min())
```

### Q10. Using the Sales column, calculate:
1. Total Sales
2. Average sales
3. Maximum sales
4. Minimum sales

```python
# Solution

print("Total Sales: ", df["Sales"].sum())
print("\nAverage Sales: ", df["Sales"].mean())
print("\nMaximum Sales: ", df["Sales"].max())
print("\nMinimum Sales: ", df["Sales"].min())
```
