### Q3. Identify the top 5 customers based on the total sales. Hint: Use GroupBy with aggregation.

```python
# Solution
top_customers = (
    df.groupby("Customer")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

display(top_customers)
```

### Q4. Calculate the total sales generated by each region. Then identify: which region generated the highest revenue?

```python
# Solution

regional_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

display(regional_sales)

highest_revenue_region = regional_sales.idxmax()

display(highest_revenue_region)
```

### Q5. Product Performance Analysis
- Calculate the total quantity sold for each product.
- Which product appears to be the most popular?

```python
# Solution Q5 --> part 1
product_quantity = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

display(product_quantity)
```

```python
# Solution Q5 --> part 2
most_popular_product = product_quantity.idxmax()
highest_quantity = product_quantity.max()

most_popular_product = pd.Series([most_popular_product, highest_quantity], index=["Product", "Quantity"])

display(most_popular_product)
```

### Q6. Multi-Level Grouping
- Group the dataset by: Region and category
- Calculate the total sales for each combination
- What insights can you derive from this analysis?

```python
# Solution Q6 --> part 1
region_category_sales = df.groupby(["Region", "Category"])["Sales"].sum().reset_index()

display(region_category_sales)
```

```python
# Solution Q6 --> part 2
region_category_sales = df.groupby(["Region", "Category"])["Sales"].sum().reset_index().sort_values("Sales", ascending=False)

display(region_category_sales)
```

###  Solution Q6 --> part 3 --> Explaination/Insights

- This analysis allows us to compare both region and category simultaneously.
- East region performs particularly well for a particular category
- Electronics & Accessories perform consistently across multiple regions
- Many region-category combinations contribute significantly more revenue than others.

### Q7. Create a pivot table showing:
- Rows --> Region
- Columns --> Product
- Values --> Total Sales

Which region sells the most laptops?

```python
# Solution
pivot_table = pd.pivot_table(
    df,
    index = "Region",
    columns = "Product",
    values = "Sales",
    aggfunc = "sum",
    fill_value = 0
)

display(pivot_table)

laptop_sales = pivot_table["Laptop"]

region_most_Laptops = laptop_sales.idxmax()
highest_laptop_sales = laptop_sales.max()

region_most_Laptops = pd.Series([region_most_Laptops, highest_laptop_sales], index=["Region", "Sales"])

display(region_most_Laptops)
```

### Q8. Profitability Analysis
- Calculate the average profit for each product category.
- Which category is most profitable?

```python
# Solution

category_profit = df.groupby("Category")["Profit"].mean().sort_values(ascending=False)

display(category_profit)

most_profitable_category = category_profit.idxmax()
highest_profit = category_profit.max()

most_profitable_category = pd.Series([most_profitable_category, highest_profit], index=["Category", "Profit"])

display(most_profitable_category)
```

### Q9. Analyze the relationship between discount and profit. Tasks:
- Calculate the average profit for each discount level
- Identify whether higher discounts lead to lower profit

```python
# Solution Q9 --> part 1
discount_profit = df.groupby("Discount")["Profit"].mean().sort_index()

display(discount_profit)
```

```python
# Solution Q9 --> part 2

correlation = df["Discount"].corr(df["Profit"])

print("Correlation is: ",correlation)
print("\nInterpretation is: ")

if correlation < 0:
  print("As discount increases, profit decreases.")
elif correlation > 0:
  print("As discount increases, profit increases.")
else:
  print("There is no correlation between discount and profit.")
```

### Q10. Perform the following workflow:
1. Filter the dataset where Sales > 3000
2. Group data by Region and Product
3. Calculate total Sales
4. Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
# Solution

filtered_df = df[df["Sales"] > 3000]
print("The dataset with Sales > 3000 is: ")
display(filtered_df)

grouped_df = filtered_df.groupby(["Region", "Product"])["Sales"].sum().reset_index()
print("\nThe grouped dataset is: ")
display(grouped_df)

final_pivot = pd.pivot_table(
    grouped_df,
    index = "Region",
    columns = "Product",
    values = "Sales",
    aggfunc = "sum",
    fill_value = 0
)
print("\nThe final pivot table is: ")
display(final_pivot)

# Interpretation
print("\nInterpretation is: ")

print("\nThe analysis focuses only on orders with sales greater than 3000. The pivot table makes it easy to compare high-value sales across different regions and products. It helps identify which products generate the highest sales in each region and which regions have stronger performance for particular products. This information can be used to identify high-performing products, strong regional markets, and potential areas for sales improvement.")
```
